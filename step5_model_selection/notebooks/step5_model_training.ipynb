That means the notebook must prove:

- why the final model is the best one,
- why the rejected models were weaker,
- why the chosen metrics are correct,

## 2. What Kind of Model Was Used?

This project uses **regression models** because the output is a continuous numerical value: `Pm2.5` concentration.

The algorithms considered in Step 5 are:

- `LinearRegression`
- `RandomForestRegressor`
- `GradientBoostingRegressor`
- `RF_Tuned`
- `ET_Tuned`
- `GB_Tuned`
- `Stacked_Ensemble`

The final selected model is not a classification model, because AQI category prediction was not the target here. The goal is direct PM2.5 value prediction.


## 3. How Was the Model Trained?

The training process followed a proper machine-learning workflow:

- train-test split: `80/20`
- validation strategy: `5-fold cross-validation`
- baseline comparison first,
- tuned advanced models second,
- final model selected using test metrics.

This is important because it prevents choosing a model based only on intuition. Instead, the final algorithm is chosen after comparison and validation.


```python
from pathlib import Path
import pandas as pd

metrics_dir = Path('../metrics')
plots_dir = Path('../plots')
models_dir = Path('../models')
docs_dir = Path('../docs')
dashboard_dir = Path('../open_source_dashboard')

baseline_comparison = pd.read_csv(metrics_dir / 'model_comparison.csv')
advanced_comparison = pd.read_csv(metrics_dir / 'tuned_model_comparison.csv')
best_metrics = pd.read_csv(metrics_dir / 'best_metrics.csv')
baseline_best = pd.read_csv(metrics_dir / 'best_model_metrics.csv')
feature_importance = pd.read_csv(metrics_dir / 'feature_importance.csv')
predictions = pd.read_csv(metrics_dir / 'predictions_vs_actual.csv')

print('Baseline comparison table')
display(baseline_comparison)

print('Advanced comparison table')
display(advanced_comparison)

print('Best metrics table')
display(best_metrics)
```

## 4. How Did the Models React?

The models reacted differently to the dataset:

- `LinearRegression` reacted poorly because the PM2.5 relationships are not purely linear. It produced a negative `R2`, which shows bad fit.
- `RandomForestRegressor` and `GradientBoostingRegressor` reacted much better because the data has nonlinear patterns and feature interactions.
- The tuned tree-based models improved performance further.
- The final `Stacked_Ensemble` reacted best because it combined strengths from multiple strong learners.

So the dataset behavior clearly favors nonlinear ensemble methods over a simple linear model.


## 5. Final Model Accuracy and Performance Metrics

The final selected model is **Stacked_Ensemble**.

Final Step 5 performance:

- `R2 Score = 0.8186`
- `MAE = 19.6332`
- `RMSE = 28.8320`

These are the correct metrics because this is a regression problem:

- `R2 Score` measures explained variance.
- `MAE` measures average absolute prediction error.
- `RMSE` measures larger prediction errors more strongly.

Together, these metrics show both fit quality and error magnitude.


```python
best_metric_map = dict(zip(best_metrics['metric'], best_metrics['value']))
baseline_metric_map = dict(zip(baseline_best['metric'], baseline_best['value']))

print('Selected final model:', best_metric_map['Selected Model'])
print('Final R2:', round(float(best_metric_map['R2 Score']), 4))
print('Final MAE:', round(float(best_metric_map['MAE']), 4))
print('Final RMSE:', round(float(best_metric_map['RMSE']), 4))

print('\nBest baseline model:', baseline_metric_map['Selected Model'])
print('Baseline R2:', round(float(baseline_metric_map['R2 Score']), 4))
print('Baseline MAE:', round(float(baseline_metric_map['MAE']), 4))
print('Baseline RMSE:', round(float(baseline_metric_map['RMSE']), 4))
```

## 6. Why This Algorithm?

The final algorithm is **Stacked_Ensemble** because it is the strongest model in the tuned comparison table.

Why it was selected:

- it achieved the best `test_r2`,
- it achieved the lowest `test_mae`,
- it achieved the lowest `test_rmse`,
- and it combines multiple strong learners instead of depending on one algorithm only.

That is a stronger justification than saying only that ensemble methods are powerful. The real reason is that the metric table proves it.


```python
print('Advanced models sorted by test R2')
display(advanced_comparison.sort_values('test_r2', ascending=False))

print('Saved model files')
for p in sorted(models_dir.glob('*.pkl')):
    print('-', p.name)
```

## 7. Why Not the Other Algorithms?

### Why not LinearRegression?
- It gave a negative `R2`, which means it could not fit the data pattern properly.
- The AQI feature relationships are nonlinear, so a simple linear boundary is not enough.

### Why not RandomForestRegressor alone?
- It is a strong baseline, but its final test metrics are still slightly weaker than the ensemble.

### Why not GradientBoostingRegressor alone?
- It is the best baseline model, but it still does not beat the final tuned ensemble.

### Why not RF_Tuned, ET_Tuned, or GB_Tuned as final model?
- All three are good models, but the final table shows that `Stacked_Ensemble` still gives the best balance of `R2`, `MAE`, and `RMSE`.

So the rejected algorithms were not bad. They were simply not the strongest according to the evaluation evidence.

## 9. Why This Tool and Why Not Metabase, Superset, or StyleBI?

### Why Streamlit + Plotly was chosen
- It is open source.
- It is easy to run locally during viva.
- It is excellent for model metrics, model ranking, residuals, and prediction diagnostics.
- It allows interactive presentation of machine-learning results.

### Why not Metabase for this step
- Metabase is good for business-style dashboarding over structured tables.
- Step 5 needs model diagnostics and prediction views, which Streamlit handles more directly.

### Why not Superset for this step
- Superset is powerful for large analytical dashboards, but it is heavier to set up and defend for a project of this size.
- For Step 5, the simpler tool is more effective because the focus is model outputs, not warehouse-scale BI.

### Why not StyleBI for this step
- StyleBI is more dashboard oriented, while this step needs custom machine-learning presentation and justification.
- Streamlit allows tighter control over model-related visual layout and explanation.

So the choice of Streamlit + Plotly is not because the other tools are bad. It is because this particular step is centered on machine-learning evidence.
