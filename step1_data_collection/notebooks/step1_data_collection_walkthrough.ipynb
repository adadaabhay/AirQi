## Scraper Flow
1. `setup_pull.py`: builds encoded query payloads for date ranges/stations.
2. `pull.py`: sends requests and stores JSON payload hashes/status.
3. `parse.py`: parses response body into table-ready rows.
4. `get_aqi.py`: Data.gov pull loop for accumulating target record count.

This flow satisfies rubric expectations for data source clarity and collection traceability.
