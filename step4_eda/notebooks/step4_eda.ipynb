## Why these visualizations only?
The selected plots are intentionally non-redundant and each answers a specific question:
1. Distribution plot: how skewed PM2.5 is.
2. Correlation heatmap: which predictors move with target.
3. Relationship plots: whether high-signal predictors have monotonic trends.
4. Outlier boxplots: whether robust models are required.
5. Category/geospatial plots: policy-facing interpretation by AQI classes and state/latitude.


```python
from pathlib import Path
import pandas as pd

stats_dir = Path('../csv')
print('Stats files:')
for p in sorted(stats_dir.glob('*.csv')):
    print('-', p.name)

summary = pd.read_csv(stats_dir / 'summary_statistics.csv', index_col=0)
target_corr = pd.read_csv(stats_dir / 'correlation_with_target.csv')
print('
Summary shape:', summary.shape)
print('
Top target correlations:')
print(target_corr.head(10))
```
